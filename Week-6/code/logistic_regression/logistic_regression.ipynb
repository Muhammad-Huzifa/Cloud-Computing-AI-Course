```python
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler , OneHotEncoder
from sklearn.metrics import r2_score
from sklearn.linear_model import LogisticRegression 
from sklearn.compose import ColumnTransformer
import numpy as np
import pandas as pd
```

```python
df = pd.read_csv("student_pass_fail.csv")

df.head()
```

```python
X = df.drop(columns=["Pass_Fail"])
y = df["Pass_Fail"]

print(X.head())
print(y.head())
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    
)

print("Training students:", len(X_train))
print("Testing students:", len(X_test))
```

```python
number_columns = ["Study_Hours_Per_Day" , "Attendance_Percentage", "Previous_Year_Score", "Sleep_Hours"]
text_columns = ["Internet_Access" , "Extracurricular_Activities"]
```

```python
preprocessor = ColumnTransformer(
    transformers=[
        (
            "numbers",
            StandardScaler(),
            number_columns
        ),
        (
            "text",
            OneHotEncoder(handle_unknown="ignore"),
            text_columns
        )
    ]
)
```

```python
from sklearn.pipeline import Pipeline
```

```python
model = Pipeline(
    steps=[
        ("preprocessing", preprocessor),
        ("classification", LogisticRegression(max_iter=1000))
    ]
)
```

```python
model.fit(X_train, y_train)

print("Model training completed")
```

```python
from sklearn.metrics import accuracy_score
```

```python
y_pred = model.predict(X_test)
```

```python
print(y_test[:5].tolist())
print(y_pred[:5])
```

```python
accuracy = accuracy_score(y_test, y_pred)

print(f"Accuracy: {accuracy * 100:.2f}%")
```

```python
import joblib
```

```python
joblib.dump(
    model,
    "student_pass_fail_model.pkl"
)

print("Model saved successfully")
```
