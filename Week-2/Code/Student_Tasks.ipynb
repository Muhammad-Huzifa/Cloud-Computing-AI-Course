# Student Tasks — Python: The Language Behind AI

Solve each task in the empty cell below it.
Use only what was covered in the lesson examples.

**How to use:**
- Read the task carefully
- Write your code in the code cell under the task
- Run the cell to check your answer

### Task 1 — Your First Program
*(Topic: What Is a Program?)*

1. Print your name on one line and your favourite subject on the next line.
2. Create `x = 8` and `y = 4`, then print their sum.

### Task 2 — Variables
*(Topic: Variables)*

1. Create three variables: `city`, `temperature` and `is_sunny`.
2. Print all three with a label.
3. Change `temperature` to a new value and print it again.

### Task 3 — Data Types
*(Topic: Data Types)*

1. Create `count = "7"` (a string) and print its type.
2. Convert it to an integer, add `3`, and print the result.
3. Print the type of the new result.

### Task 4 — Operators
*(Topic: Operators)*

1. Create `a = 25` and `b = 4`.
2. Print the result of `+`, `-`, `*`, `/`, `//`, `%` and `**` on them.
3. Print whether `a > b`.
4. Print the result of `a > 10 and b > 10`.

### Task 5 — Conditionals
*(Topic: Conditionals)*

Given `age = 16`, print:
- "Child" if age is under 13
- "Teenager" if age is under 20
- "Adult" otherwise

Then run the same code again with `age = 25`.

### Task 6 — Collections
*(Topic: Collections)*

1. Create a list `fruits = ["apple", "banana", "mango"]`. Print the first item, append `"orange"`, then print the list again.
2. Create a dict `student` with keys `name`, `roll_no` and `marks`. Print the name, update the marks, then print the whole dict.
3. Create a set from the list `["a", "b", "a", "c"]` and print it.

### Task 7 — Loops
*(Topic: Loops)*

1. Use a for loop to print the numbers 1 to 10.
2. Loop over `names = ["Ali", "Sara", "Bilal"]` and print `"Hello, {name}"` for each.
3. Use a while loop that starts at `n = 5` and keeps printing `n` while it is greater than 0, decreasing it by 1 each time.

### Task 8 — Functions
*(Topic: Functions)*

1. Write a function `square(n)` that returns `n * n`. Call it with `4` and `9`.
2. Write a function `is_pass(mark)` that returns `"Pass"` if the mark is 50 or above, otherwise `"Fail"`. Call it with `70` and `35`.

### Task 9 — Classes
*(Topic: Classes)*

Create a class `Student` with:
- `__init__(self, name)` — stores `self.name` and sets `self.marks = []`
- `add_mark(self, mark)` — appends the mark to `self.marks`
- `total(self)` — returns the sum of `self.marks`

Then create one student, add three marks, and print the name and the total.

### Task 10 — Final Task (Everything Together)
*(Topics: Classes + Loops + Collections + Conditionals)*

Create a class `ResultSheet` with:
- `__init__(self)` — initializes `self.records = []`
- `add(self, name, mark)` — appends the tuple `(name, mark)` to `self.records`
- `show(self)` — loops through the records and prints `"{name} -> Pass"` if the mark is 50 or above, otherwise `"{name} -> Fail"`

Then add three students with their marks and call `show()`.
