## 1. What Is a Program?

A program is a set of instructions a computer follows, step by step.

- Like a recipe: ingredients (data) + steps (logic) = result
- The computer does exactly what it is told — no guessing, no creativity
- Every AI model (ChatGPT, image generators) is built from programs

**Why Python for AI?**
Python is the *interface layer*. You write clear instructions, and Python calls powerful libraries that do the heavy math.
