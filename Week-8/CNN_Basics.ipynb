## 1. CNN in plain language

Think of a CNN as small windows scanning across an image. **Convolution** filters look for patterns in small areas and learn which patterns matter. **ReLU** helps the network learn complex patterns. **Pooling** shrinks feature maps while keeping strong signals. The final classifier uses those features to choose a class.

A grayscale image has shape `(height, width, 1)`, for example `(28, 28, 1)`. An RGB color image has 3 channels, such as `(128, 128, 3)`.

A Dense network flattens the image first. A CNN looks at local neighborhoods and reuses filters across the image, which helps it learn spatial patterns such as edges and shapes.

````python
import tensorflow as tf
import numpy as np
import matplotlib.pyplot as plt

print("TensorFlow version:", tf.__version__)
````

## 2. Load a small image dataset

Fashion MNIST contains 70,000 small grayscale clothing images in 10 categories. We train on the training set and keep the test set aside to check performance on unseen images. Pixel values are scaled from 0â€“255 to 0â€“1.

````python
(x_train, y_train), (x_test, y_test) = tf.keras.datasets.fashion_mnist.load_data()
class_names = ["T-shirt/top", "Trouser", "Pullover", "Dress", "Coat", "Sandal", "Shirt", "Sneaker", "Bag", "Ankle boot"]
x_train = x_train.astype("float32") / 255.0
x_test = x_test.astype("float32") / 255.0
# Conv2D expects height, width, channels. Add one channel for grayscale.
x_train = x_train[..., np.newaxis]
x_test = x_test[..., np.newaxis]
print("Training images:", x_train.shape)
print("Test images:", x_test.shape)
````

````python
plt.figure(figsize=(8, 4))
for i in range(8):
    plt.subplot(2, 4, i + 1)
    plt.imshow(x_train[i].squeeze(), cmap="gray")
    plt.title(class_names[y_train[i]])
    plt.axis("off")
plt.tight_layout()
````

## 3. Build and train the CNN

`Conv2D(32, 3)` learns 32 filters of size 3Ã—3. `MaxPooling2D(2)` roughly halves image height and width. `Flatten` prepares features for the classifier. The final `Dense(10, softmax)` returns probabilities for 10 classes.

We use Adam to update weights and sparse categorical cross-entropy because labels are integer IDs (0â€“9). An epoch is one pass through training data. Three epochs keeps the first run short.

````python
model = tf.keras.Sequential([
    tf.keras.layers.Input(shape=(28, 28, 1)),
    tf.keras.layers.Conv2D(32, 3, activation="relu"),
    tf.keras.layers.MaxPooling2D(2),
    tf.keras.layers.Conv2D(64, 3, activation="relu"),
    tf.keras.layers.MaxPooling2D(2),
    tf.keras.layers.Flatten(),
    tf.keras.layers.Dense(64, activation="relu"),
    tf.keras.layers.Dense(10, activation="softmax")
])
model.summary()
````

````python
model.compile(optimizer="adam", loss="sparse_categorical_crossentropy", metrics=["accuracy"])
history = model.fit(x_train, y_train, validation_split=0.1, epochs=3, batch_size=64)
````

````python
plt.plot(history.history["accuracy"], label="training")
plt.plot(history.history["val_accuracy"], label="validation")
plt.xlabel("Epoch"); plt.ylabel("Accuracy"); plt.legend(); plt.grid(True)
````

## 4. Evaluate and predict

The test images were not used to train the model. Test accuracy gives a better estimate of performance on new images.

````python
test_loss, test_accuracy = model.evaluate(x_test, y_test, verbose=0)
print(f"Test accuracy: {test_accuracy:.1%}")
index = 0
probabilities = model.predict(x_test[index:index + 1], verbose=0)[0]
predicted_id = int(np.argmax(probabilities))
actual_id = int(y_test[index])
plt.imshow(x_test[index].squeeze(), cmap="gray"); plt.axis("off")
plt.title(f"Predicted: {class_names[predicted_id]} | Actual: {class_names[actual_id]}")
plt.show()
print("Confidence:", f"{probabilities[predicted_id]:.1%}")
````

## 5. Use a Kaggle image dataset

Download and unzip an image-classification dataset. For a first lesson, choose a small, clearly labelled dataset with 2â€“5 categories. Arrange files into one folder per category:

```text
my_images/
    cats/

```

Folder names become labels. Keep classes reasonably balanced and begin with a few hundred images so student computers can run it. Check the dataset's terms and attribution before sharing. Extract it inside `Week-8/data/my_images/`; point `DATA_DIR` to the folder directly containing the class folders. If the ZIP has an extra parent folder, point inside it.

````python
from pathlib import Path

DATA_DIR = Path("data/my_images")  # change to your extracted dataset folder
IMAGE_SIZE = (128, 128)
BATCH_SIZE = 32

train_ds = tf.keras.utils.image_dataset_from_directory(
    DATA_DIR, validation_split=0.2, subset="training", seed=123,
    image_size=IMAGE_SIZE, batch_size=BATCH_SIZE
)
val_ds = tf.keras.utils.image_dataset_from_directory(
    DATA_DIR, validation_split=0.2, subset="validation", seed=123,
    image_size=IMAGE_SIZE, batch_size=BATCH_SIZE
)
class_names = train_ds.class_names
num_classes = len(class_names)
print("Classes:", class_names)
train_ds = train_ds.cache().shuffle(500).prefetch(tf.data.AUTOTUNE)
val_ds = val_ds.cache().prefetch(tf.data.AUTOTUNE)
````

### Train the same idea on Kaggle images

These are RGB images. `Rescaling` changes pixel values from 0â€“255 to 0â€“1. The final layer automatically matches the number of class folders.

````python
kaggle_model = tf.keras.Sequential([
    tf.keras.layers.Input(shape=(*IMAGE_SIZE, 3)),
    tf.keras.layers.Rescaling(1.0 / 255),
    tf.keras.layers.Conv2D(32, 3, activation="relu"),
    tf.keras.layers.MaxPooling2D(),
    tf.keras.layers.Conv2D(64, 3, activation="relu"),
    tf.keras.layers.MaxPooling2D(),
    tf.keras.layers.Conv2D(128, 3, activation="relu"),
    tf.keras.layers.MaxPooling2D(),
    tf.keras.layers.Flatten(),
    tf.keras.layers.Dense(128, activation="relu"),
    tf.keras.layers.Dropout(0.3),
    tf.keras.layers.Dense(num_classes, activation="softmax")
])
kaggle_model.compile(optimizer="adam", loss="sparse_categorical_crossentropy", metrics=["accuracy"])
kaggle_history = kaggle_model.fit(train_ds, validation_data=val_ds, epochs=5)
````
