# Titanic Dataset Analysis with Pandas

## About the Dataset

The Titanic dataset contains information about passengers who travelled on the Titanic. It includes each passenger's age, sex, ticket class, fare and survival status.

- **Survived = 0:** The passenger did not survive
- **Survived = 1:** The passenger survived
- **Pclass:** Passenger class: 1st, 2nd or 3rd
- **SibSp:** Number of siblings or spouses aboard
- **Parch:** Number of parents or children aboard

We will load, explore, clean, filter, group and visualize this dataset step by step.

## 1. Import the Libraries

Pandas works with tables, NumPy performs numerical calculations and Matplotlib creates charts.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

## 9. Find Missing Values

The following code counts empty values in every column.

```python
df.isnull().sum()
```

## 10. Handle Missing Values

- Replace missing Age values with the median age.
- Replace missing Fare values with the median fare.
- Replace missing Embarked values with the most common port.
- Remove Cabin because it contains many missing values.

```python
df["Age"] = df["Age"].fillna(df["Age"].median())
df["Fare"] = df["Fare"].fillna(df["Fare"].median())
df["Embarked"] = df["Embarked"].fillna(df["Embarked"].mode()[0])

df = df.drop(columns=["Cabin"])

print("Missing values handled.")
```

## 11. Check Missing Values Again

Run the same check to confirm that the important missing values are removed.

```python
df.isnull().sum()
```

## 12. Check and Remove Duplicate Records

PassengerId uniquely identifies each passenger, so repeated IDs can be removed safely.

```python
print("Duplicate rows:", df.duplicated().sum())

df = df.drop_duplicates(subset="PassengerId")

print("Duplicates removed.")
```

## 13. Select Important Columns

We can select only the columns needed for a task.

```python
selected_data = df[["PassengerId", "Name", "Sex", "Age", "Pclass", "Survived"]]

selected_data.head()
```

## 14. Select Data with loc

Loc selects rows and columns by their labels.

```python
df.loc[0:4, ["Name", "Sex", "Age", "Survived"]]
```

## 15. Select Data with iloc

Iloc selects rows and columns by their numerical positions.

```python
df.iloc[0:5, 0:4]
```

## 16. Filter Surviving Passengers

This condition keeps only passengers whose Survived value is 1.

```python
survived_passengers = df[df["Survived"] == 1]

survived_passengers[["Name", "Sex", "Age", "Pclass"]].head()
```

## 17. Filter Using Two Conditions

This example keeps female passengers who survived.

```python
female_survivors = df[(df["Sex"] == "female") & (df["Survived"] == 1)]

female_survivors[["Name", "Age", "Pclass", "Fare"]].head()
```

## 18. Sort Passengers by Fare

The highest fares will appear first.

```python
highest_fares = df.sort_values(by="Fare", ascending=False)

highest_fares[["Name", "Pclass", "Fare"]].head()
```

## 19. Create New Columns

FamilySize combines the passenger, siblings, spouse, parents and children. Status changes 0 and 1 into readable labels.

```python
df["FamilySize"] = df["SibSp"] + df["Parch"] + 1

df["Status"] = df["Survived"].replace({
    0: "Did Not Survive",
    1: "Survived"
})

df[["Name", "SibSp", "Parch", "FamilySize", "Status"]].head()
```

## 20. Calculate Survival Rate by Sex

Groupby places male and female passengers into separate groups. Mean calculates the survival rate.

```python
survival_by_sex = df.groupby("Sex")["Survived"].mean() * 100

survival_by_sex
```

## 21. Calculate Survival Rate by Passenger Class

This compares the survival percentages of first, second and third class.

```python
survival_by_class = df.groupby("Pclass")["Survived"].mean() * 100

survival_by_class
```

## 22. Analyse Age with NumPy

The Age column is converted into a NumPy array before calculating basic statistics.

```python
age_array = df["Age"].to_numpy()

print("Average age:", np.mean(age_array))
print("Median age:", np.median(age_array))
print("Youngest age:", np.min(age_array))
print("Oldest age:", np.max(age_array))
```

## 23. Visualize Passenger Outcomes

A bar chart compares passengers who survived with those who did not.

```python
survival_counts = df["Survived"].value_counts().sort_index()

plt.bar(
    ["Did Not Survive", "Survived"],
    survival_counts.values,
    color=["#E76F51", "#2A9D8F"]
)

plt.title("Titanic Passenger Outcomes")
plt.xlabel("Outcome")
plt.ylabel("Number of Passengers")
plt.show()
```

## 24. Visualize the Age Distribution

A histogram shows how passenger ages are distributed.

```python
plt.hist(df["Age"], bins=20, color="#457B9D", edgecolor="white")

plt.title("Age Distribution of Titanic Passengers")
plt.xlabel("Age")
plt.ylabel("Number of Passengers")
plt.show()
```

## 25. Visualize Survival Rate by Sex

This chart compares the survival percentages of male and female passengers.

```python
plt.bar(
    survival_by_sex.index,
    survival_by_sex.values,
    color=["#F4A261", "#264653"]
)

plt.title("Survival Rate by Sex")
plt.xlabel("Sex")
plt.ylabel("Survival Rate (%)")
plt.ylim(0, 100)
plt.show()
```

## 26. Visualize Survival Rate by Passenger Class

This chart compares survival percentages across the three passenger classes.

```python
plt.bar(
    ["First Class", "Second Class", "Third Class"],
    survival_by_class.values,
    color=["#6D597A", "#B56576", "#E56B6F"]
)

plt.title("Survival Rate by Passenger Class")
plt.xlabel("Passenger Class")
plt.ylabel("Survival Rate (%)")
plt.ylim(0, 100)
plt.show()
```
