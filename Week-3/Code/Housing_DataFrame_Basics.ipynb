# Housing Dataset with Pandas DataFrames

This notebook focuses only on the basic Pandas DataFrame operations needed to understand and explore the housing dataset.

Each row represents one house. The columns contain information such as price, area, bedrooms, bathrooms, stories and parking.

The lesson moves from loading the dataset to viewing, selecting, filtering, cleaning, grouping and saving the data. No machine-learning model is used.


## 1. Import Pandas

Pandas is used to work with data arranged in rows and columns. The short name `pd` is commonly used for Pandas.


```python
import pandas as pd
```

## 2. Load the Housing Dataset

The CSV file is read and stored in a DataFrame named `df`. This DataFrame will be used throughout the notebook.


```python
file_path = r"C:\Users\Muhammad Huzifa\Desktop\Huawei_HCCDA_AI\Cloud-Computing-AI-Course\Cloud-Computing-AI-Course\Week-3\dataset\Houses\Housing.csv"

df = pd.read_csv(file_path)

print("Dataset loaded successfully.")
```

## 3. View the First Rows

Use `head()` to quickly check the column names and the first five houses. It lets us inspect the dataset without displaying every row.


```python
df.head()
```

## 4. View the Last Rows

Use `tail()` to display the final five houses in the dataset.


```python
df.tail()
```

## 5. Check the Dataset Size

The first shape value is the number of rows, and the second is the number of columns.


```python
print("Number of rows:", df.shape[0])
print("Number of columns:", df.shape[1])
```

## 6. View the Column Names

The `columns` property shows what information is available for each house.


```python
df.columns
```

## 7. Understand the DataFrame

Use `info()` to check the column names, data types and available values. A smaller non-null count shows that a column has missing data.


```python
df.info()
```

## 8. View the Numerical Summary

Use `describe()` to see the count, average, minimum and maximum values for numerical columns such as price, area and bedrooms.


```python
df.describe()
```

## 9. Check Missing Values

Count the empty values in every column before cleaning the dataset.


```python
df.isnull().sum()
```

## 10. Check Duplicate Rows

Duplicate rows repeat the same house record and may affect the analysis.


```python
print("Duplicate rows:", df.duplicated().sum())
```

## 11. Create a Clean DataFrame

Create a new DataFrame without missing values or duplicate rows. The original `df` remains available for comparison.


```python
clean_df = df.dropna()
clean_df = clean_df.drop_duplicates()

print("Original shape:", df.shape)
print("Clean shape:", clean_df.shape)
```

## 12. Select One Column

Select the Price column when only house prices are needed.


```python
df["price"].head()
```

## 13. Select Multiple Columns

Place several column names inside a list to create a smaller DataFrame.


```python
df[["price", "area", "bedrooms"]].head()
```

## 14. Select Data with loc

`loc` selects rows and columns by their labels. Rows labelled 0 through 4 are included.


```python
df.loc[0:4, ["price", "area", "bedrooms"]]
```

## 15. Select Data with iloc

`iloc` selects rows and columns by their numerical positions. The ending position is not included.


```python
df.iloc[0:5, 0:4]
```

## 16. Filter with One Condition

Keep only houses whose area is greater than 5,000. Filtering helps us focus on records that answer a particular question.


```python
large_houses = df[df["area"] > 5000]

large_houses[["price", "area", "bedrooms"]].head()
```

## 17. Filter with Two Conditions

Keep houses that have at least three bedrooms and at least one parking space. Both conditions must be true.


```python
family_houses = df[(df["bedrooms"] >= 3) & (df["parking"] >= 1)]

family_houses[["price", "area", "bedrooms", "parking"]].head()
```

## 18. Sort Houses by Price

Sort prices from highest to lowest so the most expensive houses appear first.


```python
sorted_houses = df.sort_values(by="price", ascending=False)

sorted_houses[["price", "area", "bedrooms"]].head()
```

## 19. Calculate Basic Price Statistics

Find the average, lowest and highest house prices using simple DataFrame calculations.


```python
print("Average price:", df["price"].mean())
print("Lowest price:", df["price"].min())
print("Highest price:", df["price"].max())
```

## 20. Count Houses by Number of Bedrooms

`value_counts()` shows how many houses have each bedroom count.


```python
df["bedrooms"].value_counts().sort_index()
```

## 21. Calculate Average Price by Bedrooms

Group houses by their number of bedrooms, then calculate the average price for every group.


```python
df.groupby("bedrooms")["price"].mean()
```

## 22. Create a New Column

Create Price Per Area by dividing each house price by its area. The new column gives a simple comparison between houses of different sizes.


```python
clean_df["price_per_area"] = clean_df["price"] / clean_df["area"]

clean_df[["price", "area", "price_per_area"]].head()
```

## 23. Save the Clean DataFrame

Save the cleaned data as a new CSV file. The original Housing.csv file remains unchanged.


```python
output_path = r"C:\Users\Muhammad Huzifa\Desktop\Housing_cleaned.csv"

clean_df.to_csv(output_path, index=False)

print("Clean dataset saved successfully.")
```

## Important DataFrame Operations

| Operation | Purpose |
|---|---|
| `read_csv()` | Load a CSV file |
| `head()` | View the first rows |
| `tail()` | View the last rows |
| `shape` | Check rows and columns |
| `columns` | View column names |
| `info()` | Check data types and missing entries |
| `describe()` | Summarize numerical columns |
| `isnull().sum()` | Count missing values |
| `duplicated()` | Find duplicate rows |
| `dropna()` | Remove rows with missing values |
| `drop_duplicates()` | Remove repeated rows |
| `loc` and `iloc` | Select rows and columns |
| `sort_values()` | Sort the DataFrame |
| `value_counts()` | Count repeated categories |
| `groupby()` | Calculate results for groups |
| `to_csv()` | Save the DataFrame |
