```python
class KeypadPhone:

    def __init__(self, brand, battery_mah):
        self.brand = brand
        self.battery_mah = battery_mah

    def make_call(self, number):
        print(f"[{self.brand}] {number} Call forward")


class SmartPhone(KeypadPhone):

    def __init__(self, brand, battery_mah, camera_mp):
        super().__init__(brand, battery_mah)

        # Apna naya feature
        self.camera_mp = camera_mp

    def take_photo(self):
        print(f"{self.camera_mp}MP camera Take Hd Photo!")

    def make_call(self, number):
        print(f"[{self.brand}] {number}  WhatsApp Video Call Start")


p1 = KeypadPhone("Nokia", 1200)
p1.make_call("0300-1234567")


p2 = SmartPhone("Samsung", 5000, 108)
p2.make_call("0300-1234567")
p2.take_photo()
p2.make_call("0330-3332333")





# p2.take_photo()
# p1 = KeypadPhone("Nokia", 1200)
# p1.make_call("0300-1234567")
# # Output: [Nokia] 0300-1234567 par aam phone call lag rahi hai... Tring Tring!

# p2 = SmartPhone("Samsung", 5000, 108)
# p2.make_call("0300-1234567")
# p2.take_photo()
```

```python
import numpy as np
import pandas as pd

data = {
    "Name": ["Ali", "Sara", "Zaid", "Hira", "Usman"],
    "Age": [22, np.nan, 25, 20, np.nan],
    "City": ["Peshawar", "Lahore", np.nan, "Peshawar", "Karachi"],
    "Score": [85, 90, np.nan, 45, 70],
}

df = pd.DataFrame(data)
print(df)
```

```python
print(df["City"].value_counts())
```

```python
print(df.isna())
print(df[df["Age"].notna()])
df["City"] = df["City"].fillna("Unknown")
avg_score = df["Score"].mean()
df["Score"] = df["Score"].fillna(avg_score)
clean_df = df.dropna()
clean_df = df.dropna(subset=["Age"])
clean_cols = df.dropna(axis=1)
df_no_city = df.drop(columns=["City"])
df_cleaned = df.drop(columns=["Age", "Score"])
df_no_row0 = df.drop(index=0)
print(df.duplicated())
df_unique = df.drop_duplicates()
```
