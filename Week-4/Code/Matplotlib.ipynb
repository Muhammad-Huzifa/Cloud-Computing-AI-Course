# Drawing Charts with Python (Matplotlib)

1. **Put your data in lists** (a list is values inside `[ ]` separated by commas)
2. **Draw the chart** (`plt.plot`, `plt.bar`, `plt.hist`, `plt.pie`)
3. **Add a title and labels**, then call `plt.show()`

### Words you will see
| Word | Meaning |
|---|---|
| title | the heading of the chart |
| label | the name written on an axis |
| legend | a small box that explains which colour is which |
| axis | the horizontal (x) and vertical (y) lines of the chart |

```python
# Run this cell first. It loads the drawing tool and gives it the short name "plt".
import matplotlib.pyplot as plt
```

---
# 1. Line Plot
**Use it when:** you want to show how something **changes over time** (temperature, price, marks across exams).

## Example 1.1: One line (the simplest chart)

```python
days = [1, 2, 3, 4, 5, 6, 7]
temperature = [22, 24, 23, 26, 28, 27, 25]

plt.plot(days, temperature)

plt.title("Temperature in One Week")
plt.xlabel("Day")
plt.ylabel("Temperature (°C)")
plt.show()
```

**Try it:** change a temperature number and run again. The line moves.
**Try it:** write `plt.plot(days, temperature, color="red", marker="o")` and see what changes.

## Example 1.2: Two lines with a legend
Two students' marks over five tests. `label=` names each line, and `plt.legend()` shows the names.

```python
tests = ["Test 1", "Test 2", "Test 3", "Test 4", "Test 5"]
ali = [60, 65, 70, 72, 80]
sara = [75, 70, 78, 82, 85]

plt.plot(tests, ali, color="blue", marker="o", label="Ali")
plt.plot(tests, sara, color="red", marker="s", label="Sara")

plt.title("Marks of Two Students")
plt.xlabel("Test")
plt.ylabel("Marks")
# plt.legend()
plt.show()
```

## Example 1.3: Making the chart look better
We make the chart bigger, add a grid, set the axis range, and write a note on the highest point.

```python
months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun"]
sales = [200, 240, 260, 380, 300, 280]

plt.figure(figsize=(8, 4))                 # width 8, height 4
plt.plot(months, sales, color="green", marker="o", linestyle="--")

plt.title("Shop Sales per Month")
plt.xlabel("Month")
plt.ylabel("Sales (in thousands)")
plt.grid(True)                             # add grid lines
plt.ylim(0, 450)                           # y-axis from 0 to 450
plt.text("Apr", 395, "Best month!")        # write text at a position
plt.show()
```

# 2. Bar Chart
**Use it when:** you want to **compare different categories** (subjects, cities, products).
Each category gets one bar.

## Example 2.1: A simple bar chart

```python
subjects = ["Math", "Physics", "Chemistry", "English", "Computer"]
marks = [78, 65, 72, 85, 90]

plt.bar(subjects, marks)

plt.title("Marks in Each Subject")
plt.xlabel("Subject")
plt.ylabel("Marks")
plt.show()
```

## Example 2.2: Colours and numbers on the bars
`color=` can be one colour or a list of colours (one per bar). `plt.bar_label(...)` writes the value on top of each bar.

```python
fruits = ["Mango", "Apple", "Grapes", "Banana", "Orange"]
sold = [150, 120, 110, 90, 70]

bars = plt.bar(fruits, sold, color=["orange", "red", "purple", "gold", "darkorange"])
plt.bar_label(bars)                        # numbers on top

plt.title("Fruits Sold Today")
plt.xlabel("Fruit")
plt.ylabel("Kilograms sold")
plt.ylim(0, 180)                           # extra space above the tallest bar
plt.show()
```

**Tip:** the data above is already arranged from highest to lowest. Ordered bars are much easier to read.

## Example 2.3: Two bars for each category (grouped bars)
We draw boys and girls side by side. To place the bars side by side, we give each bar its own position:
boys sit a little **left** of 1, 2, 3, 4 and girls a little **right**.

```python
classes = ["Class A", "Class B", "Class C", "Class D"]
boys = [20, 25, 18, 22]
girls = [18, 28, 20, 24]

boys_position = [0.8, 1.8, 2.8, 3.8]     # a little left of 1, 2, 3, 4
girls_position = [1.2, 2.2, 3.2, 4.2]    # a little right of 1, 2, 3, 4

plt.bar(boys_position, boys, width=0.4, color="skyblue", label="Boys")
plt.bar(girls_position, girls, width=0.4, color="pink", label="Girls")

plt.xticks([1, 2, 3, 4], classes)          # write class names under the bars
plt.title("Number of Students in Each Class")
plt.ylabel("Students")
plt.legend()
plt.show()
```

# 3. Bar Graph (horizontal and stacked)
A *bar graph* is the same idea as a bar chart. In this section we learn **3 useful variations**.

## Example 3.1: Horizontal bars
Use `plt.barh` when the names are long, or when you want a "ranking" look.

```python
languages = ["Go", "C++", "Java", "JavaScript", "Python"]
students = [10, 22, 30, 38, 45]

plt.barh(languages, students, color="green")

plt.title("Favourite Programming Language")
plt.xlabel("Number of students")
plt.show()
```

## Example 3.2: Stacked bars
Shows the **total** and the **parts** inside it. The second set of bars starts where the first one ends (`bottom=`).

```python
quarters = ["Q1", "Q2", "Q3", "Q4"]
online = [40, 55, 60, 75]
in_store = [60, 50, 55, 45]

plt.bar(quarters, online, color="steelblue", label="Online")
plt.bar(quarters, in_store, bottom=online, color="orange", label="In store")

plt.title("Sales by Quarter")
plt.ylabel("Sales (thousands)")
plt.legend()
plt.show()
```

# 4. Histogram
**Use it when:** you have **many numbers** and want to see how they are **spread out**
(for example: how many students got 40–50, 50–60, 60–70 ...). Python groups the numbers into ranges called **bins**.

> A histogram is **not** a bar chart. A bar chart compares *categories* (Math, Physics...). A histogram shows how *one set of numbers* is spread, so the bars touch each other.

## Example 4.1: Marks of 30 students

```python
marks = [45, 52, 58, 60, 62, 65, 67, 68, 70, 72,
         55, 63, 64, 66, 69, 71, 73, 75, 77, 80,
         82, 85, 48, 59, 61, 74, 78, 88, 90, 56]

plt.hist(marks, bins=6, edgecolor="black")

plt.title("Marks of 30 Students")
plt.xlabel("Marks")
plt.ylabel("Number of students")
plt.show()
```

**Read it like this:** the tallest bar is the mark range where most students fall.

## Example 4.2: What does `bins` do?
`bins` is the number of ranges. We draw the same data 3 times (`plt.subplot(1, 3, 1)` means: 1 row, 3 charts, this is chart number 1).

```python
marks = [45, 52, 58, 60, 62, 65, 67, 68, 70, 72,
         55, 63, 64, 66, 69, 71, 73, 75, 77, 80,
         82, 85, 48, 59, 61, 74, 78, 88, 90, 56]

plt.figure(figsize=(12, 3.5))

plt.subplot(1, 3, 1)
plt.hist(marks, bins=3, edgecolor="black", color="lightblue")
plt.title("3 bins (too few)")

plt.subplot(1, 3, 2)
plt.hist(marks, bins=6, edgecolor="black", color="lightgreen")
plt.title("6 bins (good)")

plt.subplot(1, 3, 3)
plt.hist(marks, bins=20, edgecolor="black", color="salmon")
plt.title("20 bins (too many)")

plt.tight_layout()
plt.show()
```

**Lesson:** too few bins hide the pattern, too many bins make it messy. Try a few values and pick the clearest.

## Example 4.3: Compare two groups and show the average
`alpha=` makes the colour see-through so both groups can be seen. `plt.axvline` draws a vertical line at the average.

```python
import matplotlib.pyplot as plt
section_a = [50, 55, 52, 58, 60, 62, 57, 54, 59, 61, 63, 56, 65, 53, 60]
section_b = [65, 70, 72, 68, 75, 78, 74, 71, 69, 80, 73, 77, 66, 76, 79]

average_a = sum(section_a) / len(section_a)
average_b = sum(section_b) / len(section_b)
print(average_a)
plt.hist(section_a, bins=8, alpha=0.6, color="blue", label="Section A")
plt.hist(section_b, bins=8, alpha=0.6, color="orange", label="Section B")
plt.axvline(average_a, color="blue", linestyle="--")
plt.axvline(average_b, color="orange", linestyle="--")

plt.title("Marks: Section A vs Section B (dashed line = average)")
plt.xlabel("Marks")
plt.ylabel("Number of students")
plt.legend()
plt.show()
```

# 5. Pie Chart
**Use it when:** you want to show **parts of a whole** (every slice is a share of 100%).

> **Good practice:** use a pie chart only when you have **5 slices or fewer**. With many slices, a bar chart is easier to read.

## Example 5.1: A simple pie chart
`autopct="%1.1f%%"` prints the percentage on every slice.

```python
expenses = ["Rent", "Food", "Transport", "Savings", "Other"]
amount = [35, 25, 15, 15, 10]

plt.pie(amount, labels=expenses, autopct="%1.1f%%")

plt.title("My Monthly Budget")
plt.show()
```

## Example 5.2: Colours, pulled-out slice and start angle
`explode` pulls a slice out (one number per slice). `startangle=90` starts the first slice at the top.

```python
subjects = ["Python", "Java", "C++", "Other"]
share = [50, 25, 15, 10]

plt.pie(share,
        labels=subjects,
        colors=["steelblue", "orange", "green", "red"],
        explode=[0.1, 0, 0, 0],
        autopct="%1.0f%%",
        startangle=90)

plt.title("Languages Used in Class Projects")
plt.show()
```

## Example 5.3: Donut chart
A donut is a pie with a hole. `wedgeprops={"width": 0.4}` makes the hole, and `plt.text` writes in the middle.

```python
status = ["Present", "Absent", "Late", "Leave"]
count = [210, 15, 25, 10]

plt.pie(count,
        labels=status,
        autopct="%1.1f%%",
        startangle=90,
        wedgeprops={"width": 0.4, "edgecolor": "white"})

plt.text(0, 0, "260\nrecords", ha="center", va="center", fontsize=14)
plt.title("Attendance Summary")
plt.show()
```

# Quick Reference
| Chart | Command | Use it for |
|---|---|---|
| Line | `plt.plot(x, y)` | change over time |
| Bar | `plt.bar(names, values)` | compare categories |
| Horizontal bar | `plt.barh(names, values)` | long names, rankings |
| Histogram | `plt.hist(numbers, bins=6)` | spread of many numbers |
| Pie | `plt.pie(values, labels=names)` | parts of a whole |

**Always add:** `plt.title(...)`, `plt.xlabel(...)`, `plt.ylabel(...)`, and finish with `plt.show()`.
